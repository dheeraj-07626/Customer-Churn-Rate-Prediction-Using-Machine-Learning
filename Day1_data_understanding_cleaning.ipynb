**step1-import required libraries**

```python
import pandas as pd
import numpy as np
from google.colab import files
```

**step2-upload the dataset**

```python
uploaded=files.upload()
filename=next(iter(uploaded))
df=pd.read_csv(filename)

print("dataset loaded successfully")
print("rows and columns:",df.shape)
display(df.head())
```

**step3-understand the dataset**

```python
print("First 5 rows:")
display(df.head())
print("Last 5 rows:")
display(df.tail())
print("dataset shape:",df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types and non-null counts:")
df.info()
print("\nNumerical summary:")
display(df.describe().T)
```

**step4-check missing values and duplicates**

```python
missing=pd.DataFrame({
    "Missing_Count":df.isna().sum(),
    "Missing_percentage":(df.isna().mean()*100).round(2)
}).sort_values("Missing_Count",ascending=False)

display(missing)
print("exact duplicated rows:",df.duplicated().sum())

if "Customer_ID" in df.columns:
  print("Duplicate Customer_ID values:",df["Customer_ID"].duplicated().sum())
```

**step5-check unsual values**

```python
for col in ["Monthly_Charges", "Total_Charges"]:
  if col in df.columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    print(f"\n{col} summary:")
    display(df[col].describe(percentiles=[.5, .75, .95, .99]).to_frame())
    print(f"values above 10,000 in {col}:", (df[col] > 10000).sum())
```

**step6-clean columns names,text and placeholders**

```python
clean_df=df.copy()

clean_df.columns=clean_df.columns.str.strip()

placeholder_values=["","N/A","NA","NULL","None","Not Available","-",
"Unknown"]
for col in clean_df.select_dtypes(include=["object"]).columns:
  clean_df[col]=clean_df[col].astype("string").str.strip()
  clean_df[col]=clean_df[col].replace(placeholder_values,pd.NA)

print("Text cleanup completed.")
```

**step7-remove duplicate records and standardize customer IDs**

```python
before = len(clean_df)
clean_df = clean_df.drop_duplicates().copy()
if "Customer_ID" in clean_df.columns:
  clean_df["Customer_ID"] = (
      clean_df["Customer_ID"].astype("string").str.upper()
      .str.replace("_", "-", regex=False).str.strip()
  )
  clean_df = clean_df.drop_duplicates(subset=["Customer_ID"], keep="first").copy()

print("Rows before duplicated removal:", before)
print("Rows after duplicated removal:", len(clean_df))
```

**step8-correct data types and handle suspicious charge values**

```python
numeric_cols=[
    "Age","Tenure_Months","Monthly_Charges","Total_Charges",
    "Support_calls","complaints","Internet_Usage_GB",
    "Avg_session_Minutes","Discount","Satisfaction_score",
    "Customer_service_rating","Churn","Signup_year","Signup_month"
]
for col in numeric_cols:
  if col in clean_df.columns:
    clean_df[col]=pd.to_numeric(clean_df[col],errors="coerce")
if"Monthly_charges" in clean_df.columns:
  clean_df.loc[clean_df["Monthly_Charges"]>10000,"Monthly_charges"]=np.nan
  clean_df["Monthly_charges"]=clean_df["Monthly_Charges"].fillina(
      clean_df["Monthly_Charges"].median()
  )
  if"Total_charges" in clean_df.columns:
    clean_df.loc[clean_df["Total_charges"]<0,"Total_charges"]=np.nan
    clean_df.loc[clean_df["Total_charges"]>1000000,"Total_charges"]=np.nan
  if{"Total_charges","Monthly_charges","Tenure"}.issubset(clean_df.columns):
    estimated_total_charges=clean_df["Monthly_charges"]*clean_df["Tenure"]
    clean_df["Total_charges"]=clean_df["Total_charges"].fillna(estimated_total)

  for col in clean_df.select_dtypes(include=["number"]).columns:
      if col !="Churn":
        clean_df[col]=clean_df[col].fillna(clean_df[col].median())
  for col in clean_df.select_dtypes(include-["Object","string"]).columns:
      clean_df[col]=clean_df[col].fillna("unknown")
  print("cleaning completed.")

```

**step9-final validaion**

```python
print("Final dataset shape:", clean_df.shape)
print("total missing values:", clean_df.isna().sum().sum())
print("total duplicated rows:", clean_df.duplicated().sum())

if "Churn" in clean_df.columns:
  print("\nchurn values:")
  print(clean_df["Churn"].value_counts(dropna=False))
display(clean_df.head())
display(clean_df.describe(include="all").T)
```

**step10-save and download the cleaned dataset**

```python
cleaned_file="customer_churn_cleaned.csv"
clean_df.to_csv(cleaned_file,index=False)

print("Saved:",cleaned_file)
files.download(cleaned_file)
```
